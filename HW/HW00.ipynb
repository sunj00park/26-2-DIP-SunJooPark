```python
from PIL import Image
from PIL.ExifTags import TAGS
```

```python
I = Image.open('/content/sample_data/IMG_5002.jpeg')
exifdata = I.getexif()
for tag_id in exifdata:
    tag = TAGS.get(tag_id, tag_id)
    data = exifdata.get(tag_id)
    print(f"{tag:25}: {data}")
```
